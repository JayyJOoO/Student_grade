```python
import io
import pandas as pd
from google.colab import files

# 1. Trigger the file upload prompt
uploaded = files.upload()

# 2. Dynamically get the name of the uploaded file
if uploaded:
    file_name = list(uploaded.keys())[0]
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))

    # 3. View the data
    display(df.head(10))
else:
    print('No file was uploaded.')
```

```python
import io
import pandas as pd
import ipywidgets as widgets
from IPython.display import display

# Create UI Widgets
file_upload = widgets.FileUpload(accept='.csv', multiple=False, description='Upload CSV')
output_area = widgets.Output()

def on_upload_change(change):
    output_area.clear_output()
    with output_area:
        if not file_upload.value:
            print("No file uploaded.")
            return

        # Get uploaded file information
        uploaded_file = list(file_upload.value.values(10))[0]
        content = uploaded_file['content']
        file_name = uploaded_file['metadata']['name']

        print(f"File uploaded: {file_name}")

        # Read and display DataFrame
        df = pd.read_csv(io.BytesIO(content))
        display(df.head(11))

# Bind event handler to widget
file_upload.observe(on_upload_change, names='value')

# Display GUI
display(file_upload, output_area)
```

```python
import io
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, HTML

# Cyber-Luxe White & Violet CSS Theme - Fixed Column Overlap & Dynamic Widths
css_style = """
<style>
    @import url('https://fonts.googleapis.com/css2?family=Orbitron:wght@500;700;900&family=Fira+Code:wght@400;600&display=swap');

    /* Global Card Container */
    .report-card {
        background: #ffffff !important;
        border: 2px solid #8a2be2 !important;
        border-radius: 12px !important;
        padding: 24px !important;
        box-shadow:
            0 0 15px rgba(138, 43, 226, 0.25),
            0 0 30px rgba(147, 112, 219, 0.15) !important;
        max-width: 800px;
        margin: 15px auto;
        font-family: 'Fira Code', monospace !important;
    }

    /* Header Title */
    .report-header-title {
        font-family: 'Orbitron', sans-serif;
        color: #8a2be2;
        font-size: 20px;
        font-weight: 900;
        text-align: center;
        letter-spacing: 1.5px;
        text-transform: uppercase;
        margin-bottom: 16px;
    }

    /* Student Info Meta Section */
    .student-info-grid {
        display: grid;
        grid-template-columns: 140px 1fr;
        row-gap: 4px;
        margin-bottom: 16px;
        font-size: 13px;
    }

    .info-label {
        color: #7b1fa2;
        font-weight: 700;
        text-align: left;
    }

    .info-value {
        color: #5c068c;
        font-weight: 600;
        text-align: left;
    }

    /* Section Headers */
    .section-title {
        font-family: 'Orbitron', sans-serif;
        color: #8a2be2;
        font-size: 12px;
        font-weight: 700;
        letter-spacing: 1px;
        text-transform: uppercase;
        margin: 14px 0 8px 0;
    }

    /* Table Container - Auto Layout prevents text overlaps */
    .cyber-table-wrapper {
        border-radius: 8px;
        border: 1px solid #d8bfd8;
        background-color: #ffffff !important;
        overflow-x: auto;
    }

    .cyber-table table {
        border-collapse: collapse !important;
        width: 100% !important;
        table-layout: auto !important;
        background-color: #ffffff !important;
        font-size: 12px !important;
        line-height: 1.3 !important;
    }

    .cyber-table th {
        background: #ffffff !important;
        color: #7b1fa2 !important;
        font-family: 'Orbitron', sans-serif !important;
        font-size: 11px !important;
        letter-spacing: 0.5px !important;
        border-bottom: 2px solid #8a2be2 !important;
        padding: 8px 10px !important;
        white-space: nowrap !important;
    }

    .cyber-table td {
        color: #5c068c !important;
        font-weight: 600 !important;
        border-bottom: 1px solid #f0d5ff !important;
        background-color: #ffffff !important;
        padding: 6px 10px !important;
    }

    /* Precise Alignment Per Column */
    .cyber-table th:nth-child(1), .cyber-table td:nth-child(1) {
        text-align: left !important;
        white-space: nowrap !important;
    }

    .cyber-table th:nth-child(2), .cyber-table td:nth-child(2) {
        text-align: left !important;
        word-break: break-word !important;
    }

    .cyber-table th:nth-child(3), .cyber-table td:nth-child(3) {
        text-align: center !important;
        white-space: nowrap !important;
    }

    .cyber-table th:nth-child(4), .cyber-table td:nth-child(4) {
        text-align: right !important;
        white-space: nowrap !important;
    }

    /* Summary Box */
    .summary-box {
        background: #ffffff !important;
        border: 1px dashed #8a2be2;
        border-radius: 8px;
        padding: 10px 14px;
        margin-top: 10px;
        font-size: 13px;
    }

    .summary-row {
        display: flex;
        justify-content: space-between;
        padding: 3px 0;
        color: #5c068c;
    }

    .summary-row span {
        text-align: left;
    }

    .summary-row b {
        text-align: right;
    }

    .gwa-highlight {
        background: linear-gradient(135deg, #8a2be2, #9370db);
        color: #ffffff !important;
        font-family: 'Orbitron', sans-serif;
        font-weight: 700;
        border-radius: 6px;
        padding: 8px 12px;
        margin-top: 6px;
    }

    /* Interactive UI Controls Styling */
    .widget-text input, .widget-dropdown select {
        background-color: #ffffff !important;
        border: 1px solid #8a2be2 !important;
        border-radius: 6px !important;
        color: #5c068c !important;
        font-weight: 600 !important;
    }

    .cyber-upload > button {
        background: #ffffff !important;
        color: #8a2be2 !important;
        border: 2px solid #8a2be2 !important;
        border-radius: 8px !important;
        font-family: 'Orbitron', sans-serif !important;
        font-weight: 700 !important;
        font-size: 12px !important;
        padding: 6px 16px !important;
    }

    .generate-btn {
        background: linear-gradient(135deg, #8a2be2, #9370db) !important;
        color: #ffffff !important;
        border: none !important;
        border-radius: 8px !important;
        font-family: 'Orbitron', sans-serif !important;
        font-weight: 700 !important;
        font-size: 12px !important;
        padding: 8px 18px !important;
        cursor: pointer !important;
    }

    .print-pdf-btn {
        background: linear-gradient(135deg, #6a0dad, #8a2be2);
        color: #ffffff;
        border: none;
        border-radius: 8px;
        font-family: 'Orbitron', sans-serif;
        font-weight: 700;
        font-size: 12px;
        padding: 10px 20px;
        cursor: pointer;
        margin-top: 14px;
        display: block;
        width: 100%;
    }

    /* Print Stylesheet */
    @media print {
        @page {
            size: A4 portrait;
            margin: 10mm;
        }

        body * {
            visibility: hidden !important;
        }

        #printable-report, #printable-report * {
            visibility: visible !important;
        }

        #printable-report {
            position: absolute !important;
            left: 0 !important;
            top: 0 !important;
            width: 100% !important;
            max-width: 100% !important;
            margin: 0 !important;
            padding: 0 !important;
            border: none !important;
            box-shadow: none !important;
            background: #ffffff !important;
        }

        .print-pdf-btn {
            display: none !important;
        }
    }
</style>
"""

# Apply Stylesheet
display(HTML(css_style))

# Form Controls
file_upload = widgets.FileUpload(accept='.csv', multiple=False, description='Upload CSV')
file_upload.add_class('cyber-upload')

name_input = widgets.Text(description='Student Name:', value='Constantinopla, Jayan')

generate_button = widgets.Button(description='Generate Report')
generate_button.add_class('generate-btn')

form_title = widgets.HTML('<div style="font-family: Orbitron; color: #8a2be2; font-size: 16px; font-weight: 700; margin-bottom: 8px;">🦊 NDMU Student Report Card Generator 🦊</div>')

input_form = widgets.VBox([
    form_title,
    file_upload,
    name_input,
    widgets.Box([generate_button], layout=widgets.Layout(margin='8px 0 0 0'))
])

output_area = widgets.Output()

def parse_multi_student_csv(file_stream):
    df_raw = pd.read_csv(file_stream, encoding='latin1', header=None)
    students_data = {}
    current_name = None
    current_course_year = None
    current_rows = []

    for idx, row in df_raw.iterrows():
        val0 = str(row[0]).strip()
        if val0.startswith('Name:'):
            if current_name:
                students_data[current_name] = {
                    'course_year': current_course_year,
                    'df': pd.DataFrame(current_rows)
                }
            current_name = str(row[1]).strip()
            current_course_year = str(row[2]).strip()
            current_rows = []
        elif current_name is not None and val0 != 'Code':
            current_rows.append(row)

    if current_name and current_rows:
        students_data[current_name] = {
            'course_year': current_course_year,
            'df': pd.DataFrame(current_rows)
        }
    return students_data

def generate_report(b):
    output_area.clear_output()
    with output_area:
        try:
            if not file_upload.value:
                # Fallback to local test file
                with open('Constantinopla_Student_Gradesws.csv', 'rb') as f:
                    file_stream = io.BytesIO(f.read())
            else:
                uploaded_file = list(file_upload.value.values())[0] if isinstance(file_upload.value, dict) else file_upload.value[0]
                content = uploaded_file.get('content', b'') if isinstance(uploaded_file, dict) else uploaded_file['content']
                file_stream = io.BytesIO(content)

            students_dict = parse_multi_student_csv(file_stream)
            target_name = name_input.value.strip()

            # Find matching student (case-insensitive partial or exact match)
            matched_key = next((k for k in students_dict.keys() if target_name.lower() in k.lower()), None)

            if not matched_key:
                print(f"Student '{target_name}' not found in CSV. Available students: {list(students_dict.keys())}")
                return

            student_info = students_dict[matched_key]
            df = student_info['df']
            df.columns = ['Code', 'Teacher', 'Units', 'Grades']

            # Extract course & year level
            course_year_str = student_info['course_year']
            if '-' in course_year_str:
                parts = course_year_str.rsplit('-', 1)
                course_val = parts[0]
                year_val = parts[1]
            else:
                course_val = course_year_str
                year_val = '1'

            units_col = 'Units'
            grade_col = 'Grades'

            total_subjects = len(df)
            total_units = 0.0
            simple_avg = 0.0
            gwa = 0.0

            if grade_col in df.columns:
                numeric_grades = pd.to_numeric(df[grade_col], errors='coerce')
                simple_avg = numeric_grades.mean()

                if units_col in df.columns:
                    numeric_units = pd.to_numeric(df[units_col], errors='coerce')
                    total_units = numeric_units.sum()

                    valid_mask = numeric_grades.notna() & numeric_units.notna()
                    if numeric_units[valid_mask].sum() > 0:
                        gwa = (numeric_grades[valid_mask] * numeric_units[valid_mask]).sum() / numeric_units[valid_mask].sum()

            display(HTML(f"""
                <div id="printable-report" class="report-card">
                    <div class="report-header-title">
                        🦊 NDMU STUDENT REPORT CARD 🦊
                    </div>

                    <div class="student-info-grid">
                        <span class="info-label">Student Name:</span>
                        <span class="info-value">{matched_key}</span>
                        <span class="info-label">Course/Program:</span>
                        <span class="info-value">{course_val}</span>
                        <span class="info-label">Year Level:</span>
                        <span class="info-value">{year_val}</span>
                    </div>

                    <div class="section-title">
                        🦊 DETAILED SUBJECT GRADES
                    </div>

                    <div class="cyber-table-wrapper">
                        <div class="cyber-table">
                            {df.to_html(classes='cyber-table', index=False)}
                        </div>
                    </div>

                    <div class="section-title">
                        🦊 GRADE SUMMARY
                    </div>

                    <div class="summary-box">
                        <div class="summary-row">
                            <span>Total Subjects Taken:</span>
                            <b>{total_subjects}</b>
                        </div>
                        <div class="summary-row">
                            <span>Total Units Earned:</span>
                            <b>{total_units:.1f}</b>
                        </div>
                        <div class="summary-row">
                            <span>Simple Average Grade:</span>
                            <b>{simple_avg:.2f}</b>
                        </div>
                        <div class="summary-row gwa-highlight">
                            <span>General Weighted Average (GWA):</span>
                            <b>{gwa:.4f}</b>
                        </div>
                    </div>

                    <button class="print-pdf-btn" onclick="window.print()">🖨️ Print Report to PDF</button>
                </div>
            """))
        except Exception as e:
            print(f"Error generating report: {e}")

generate_button.on_click(generate_report)

display(input_form, output_area)
```
